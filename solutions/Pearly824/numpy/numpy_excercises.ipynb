```python
import numpy as np
```

```python
# Exercise 01 — Array Creation

# 1 1D array of 20 evenly spaced values between 0 and 1
linspace_array = np.linspace(0, 1, 20)

# 2 5x5 identity matrix
identity_matrix = np.eye(5)

# 3 3x4 matrix of random integers between 1 and 100
rand_ints = np.random.randint(1, 101, (3, 4))

# Print shapes and dtypes
print("linspace  → shape:", linspace_array.shape, " dtype:", linspace_array.dtype)
print("identity  → shape:", identity_matrix.shape, " dtype:", identity_matrix.dtype)
print("rand_ints → shape:", rand_ints.shape, " dtype:", rand_ints.dtype)
```

```python
# Exercise 02 — Reshaping and Flattening

# Step 1: Create 1D array from 1 to 24
arr = np.arange(1, 25)

print("Original  :", arr.shape)

# Step 2: Reshape to 3D (2, 3, 4)
arr_3d = arr.reshape(2, 3, 4)
print("3-D       :", arr_3d.shape)

# Step 3: Reshape to 2D (6, 4)
arr_2d = arr.reshape(6, 4)
print("2-D       :", arr_2d.shape)

# Step 4: Flatten back to 1D
arr_flat = arr_2d.flatten()
print("Flattened :", arr_flat.shape)
```

```python
#Excercise 03: Fancy Indexing & Boolean Masking


matrix = np.random.rand(6, 6)

every_other = matrix[0::2]
diag = np.diag(matrix)

binary = np.where(matrix > 0.7, 1, 0)

print("Every-other rows → shape:", every_other.shape)
print("Diagonal         → shape:", diag.shape)
print("Binary matrix    → shape:", binary.shape, " values:", np.unique(binary))
```

```python
#Excercise 04: Slicing, Stacking & Splitting

A = np.random.rand(4, 3)
B = np.random.rand(4, 3)

vstacked = np.vstack((A, B))
hstacked = np.hstack((A, B))

split = np.vsplit(vstacked, 2)

print("vstack → shape:", vstacked.shape)
print("hstack → shape:", hstacked.shape)
print("split[0] equals A:", np.array_equal(split[0], A))
print("split[1] equals B:", np.array_equal(split[1], B))
```

```python
#Excercise 05: Broadcasting Intuition


col = np.arange(4).reshape(4, 1)
row = np.arange(5).reshape(1, 5)

result1 = col + row

mat = np.arange(12).reshape(4, 3)
vec = np.arange(3)

result2 = mat + vec

print("(4,1) + (1,5)  → shape:", result1.shape)
print("(4,3) + (3,)   → shape:", result2.shape)
```

```python
#Excercise 06: Vectorisation Speed Benchmark

import time

arr = np.random.rand(1_000_000)

# Loop
start = time.time()
sqrt_loop = np.zeros_like(arr)
for i in range(len(arr)):
    sqrt_loop[i] = arr[i] ** 0.5
loop_time = time.time() - start

# NumPy
start = time.time()
sqrt_numpy = np.sqrt(arr)
numpy_time = time.time() - start

speedup = loop_time / numpy_time

print("Loop time    :", loop_time)
print("NumPy time   :", numpy_time)
print("Speedup      :", speedup, "x")
```

```python
#Excercise 07: Z-Score Normalization

data = np.random.randn(100, 5)

mean = np.mean(data, axis=0)
std = np.std(data, axis=0)

normalized = (data - mean) / std

print("Column means:", np.round(np.mean(normalized, axis=0), 2))
print("Column stds :", np.round(np.std(normalized, axis=0), 2))
```

```python
#Excercise 08: Min-Max Scaling

data = np.random.randn(100, 5)

min_vals = np.min(data, axis=0)
max_vals = np.max(data, axis=0)

scaled = (data - min_vals) / (max_vals - min_vals)

print("Min per column:", np.round(np.min(scaled, axis=0), 2))
print("Max per column:", np.round(np.max(scaled, axis=0), 2))
```

```python
#Excercise 09: Neural Network Linear Layer

W = np.random.randn(4, 3)
X = np.random.randn(3, 10)

Z = W @ X

b = np.random.randn(4, 1)
Z_bias = Z + b

print("W shape :", W.shape)
print("X shape :", X.shape)
print("Z shape :", Z.shape)
print("Z+b shape:", Z_bias.shape)
```

```python
#Excercise 10: Cosine Similarity

# Two vectors
v1 = np.random.randn(128)
v2 = np.random.randn(128)

dot = np.dot(v1, v2)
norm1 = np.linalg.norm(v1)
norm2 = np.linalg.norm(v2)

cosine = dot / (norm1 * norm2)

print("Cosine similarity (two vectors):", cosine)

# Batch
X = np.random.randn(10, 128)

X_norm = X / np.linalg.norm(X, axis=1, keepdims=True)

cosine_matrix = X_norm @ X_norm.T

print("Pairwise matrix shape          :", cosine_matrix.shape)
print("Diagonal values                :", np.round(np.diag(cosine_matrix), 2))
```
