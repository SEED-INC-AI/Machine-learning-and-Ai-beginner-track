```python
import numpy as np

# Create a 1-D array of integers from 1 to 24
arr_1d = np.arange(1, 25)
print("Original 1-D array:")
print(arr_1d)
print(f"Shape: {arr_1d.shape}\n")
print("=" * 70)

# Reshape to (2, 3, 4)
print("\n### Reshape to (2, 3, 4) ###")
arr_2_3_4 = arr_1d.reshape(2, 3, 4)
print(arr_2_3_4)
print(f"Shape: {arr_2_3_4.shape}")
print("\nML Scenario Interpretation:")
print("  - Dimension 0 (size 2): 2 batches of data")
print("  - Dimension 1 (size 3): 3 timesteps (e.g., sequence length in RNN/LSTM)")
print("  - Dimension 2 (size 4): 4 features per timestep")
print("  Example: 2 sequences, each with 3 time steps, 4 features each")
print("=" * 70)

# Reshape to (6, 4)
print("\n### Reshape to (6, 4) ###")
arr_6_4 = arr_2_3_4.reshape(6, 4)
print(arr_6_4)
print(f"Shape: {arr_6_4.shape}")
print("\nML Scenario Interpretation:")
print("  - Dimension 0 (size 6): 6 samples")
print("  - Dimension 1 (size 4): 4 features")
print("  Example: 6 data points with 4 features each (typical tabular data)")
print("=" * 70)

# Flatten back to 1-D
print("\n### Flatten back to 1-D ###")
arr_1d_final = arr_6_4.flatten()
print(arr_1d_final)
print(f"Shape: {arr_1d_final.shape}")
print("\nML Scenario Interpretation:")
print("  - Single 1-D array: Can be used as input to fully connected layers")
print("  - Or serialized for storage/transmission")
print("=" * 70)


print("\n### Verification ###")
print(f"Original and final arrays are equal: {np.array_equal(arr_1d, arr_1d_final)}")
print(f"Total elements preserved: {arr_1d.size} -> {arr_1d_final.size}")
```

```python
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Create two (4, 3) matrices of random values
matrix_1 = np.random.rand(4, 3)
matrix_2 = np.random.rand(4, 3)

print("Original Matrix 1:")
print(matrix_1)
print("\nOriginal Matrix 2:")
print(matrix_2)
print()

# Stack them vertically with np.vstack
vstack_result = np.vstack((matrix_1, matrix_2))
print("Vertically stacked result (shape: (8, 3)):")
print(vstack_result)
print(f"Shape: {vstack_result.shape}")
print()

# Stack them horizontally with np.hstack
hstack_result = np.hstack((matrix_1, matrix_2))
print("Horizontally stacked result (shape: (4, 6)):")
print(hstack_result)
print(f"Shape: {hstack_result.shape}")
print()

# Split the vertical result back into two equal arrays using np.vsplit
split_result = np.vsplit(vstack_result, 2)
split_matrix_1 = split_result[0]
split_matrix_2 = split_result[1]

print("First matrix from vertical split:")
print(split_matrix_1)
print("\nSecond matrix from vertical split:")
print(split_matrix_2)
print()

# Assert that the split arrays match the originals
print("Performing assertions...")
try:
    assert np.allclose(split_matrix_1, matrix_1), "First split matrix doesn't match original matrix 1"
    print("✓ First split matrix matches original matrix 1")
    
    assert np.allclose(split_matrix_2, matrix_2), "Second split matrix doesn't match original matrix 2"
    print("✓ Second split matrix matches original matrix 2")
    
    print("\n✓ All assertions passed! Split matrices match the originals.")
except AssertionError as e:
    print(f"✗ Assertion failed: {e}")
```

```python
import numpy as np

# Create (4, 1) column vector and (1, 5) row vector
col = np.array([[1], [2], [3], [4]])
row = np.array([[1, 2, 3, 4, 5]])

result1 = col + row
print(f"(4,1) + (1,5) = {result1.shape}")

# Add (3,) array to (4, 3) matrix
arr = np.array([10, 20, 30])
matrix = np.ones((4, 3))
result2 = matrix + arr
print(f"(4,3) + (3,) = {result2.shape}")

# Broadcasting explanation:
# NumPy broadcasts smaller arrays to match larger ones along dimensions
# (4,1) + (1,5) → (4,5): column broadcasts across columns, row across rows
# (4,3) + (3,) → (4,3): (3,) is treated as (1,3) and broadcasts down to 4 rows
```

```python
import numpy as np
import time

# Generate 1,000,000 random floats
data = np.random.rand(1_000_000)

# Approach (a): Element-by-element with Python for-loop
start = time.time()
result_loop = [x ** 0.5 for x in data]
time_loop = time.time() - start

# Approach (b): NumPy sqrt
start = time.time()
result_numpy = np.sqrt(data)
time_numpy = time.time() - start

# Calculate speedup
speedup = time_loop / time_numpy

print(f"Python for-loop:     {time_loop:.4f} seconds")
print(f"NumPy np.sqrt():     {time_numpy:.4f} seconds")
print(f"Speedup factor:      {speedup:.1f}x faster")
print(f"\nNumPy is {speedup:.0f} times faster than Python loops!")
```

```python
import numpy as np

np.random.seed(42)

# Create (100, 5) matrix: 100 samples, 5 features
data = np.random.randn(100, 5) * 10 + 50  # Random data with mean ~50, std ~10

print("Original data (first 5 rows):")
print(data[:5])
print(f"\nOriginal means: {data.mean(axis=0)}")
print(f"Original stds:  {data.std(axis=0)}")

# Normalize: (X - mean) / std
mean = data.mean(axis=0)
std = data.std(axis=0)
normalized = (data - mean) / std

print("\n" + "="*50)
print("Normalized data (first 5 rows):")
print(normalized[:5])
print(f"\nNormalized means: {normalized.mean(axis=0)}")
print(f"Normalized stds:  {normalized.std(axis=0)}")

# Verify
assert np.allclose(normalized.mean(axis=0), 0, atol=1e-10)
assert np.allclose(normalized.std(axis=0), 1, atol=1e-10)
print("\n✓ Verification passed! Means ≈ 0 and stds ≈ 1")
```

```python
import numpy as np

np.random.seed(42)

# Create (100, 5) matrix
data = np.random.randn(100, 5) * 10 + 50

# Min-max scaling: (X - min) / (max - min)
min_val = data.min(axis=0)
max_val = data.max(axis=0)
scaled = (data - min_val) / (max_val - min_val)

print(f"Scaled min: {scaled.min(axis=0)}")
print(f"Scaled max: {scaled.max(axis=0)}")

# Test point outside training range
test_point = np.array([[100, 100, 100, 100, 100]])
test_scaled = (test_point - min_val) / (max_val - min_val)

print(f"\nTest point (outside range): {test_point[0]}")
print(f"Scaled test point: {test_scaled[0]}")

# Comment: A test point outside the training range will NOT stay in [0, 1].
# Values > max scale to > 1, values < min scale to < 0.
# This is a limitation of min-max scaling: it doesn't generalize to unseen ranges.
# Standardization (z-score) is more robust for out-of-range test points.
```

```python
import numpy as np

np.random.seed(42)

# Create weight matrix W (4, 3) and input X (3, 10)
W = np.random.randn(4, 3)
X = np.random.randn(3, 10)

# Forward pass: Z = W @ X
Z = W @ X
print(f"Z shape: {Z.shape}")
assert Z.shape == (4, 10), "Output shape mismatch!"

# Add bias vector (4, 1) using broadcasting
b = np.random.randn(4, 1)
output = Z + b

print(f"Output shape (with bias): {output.shape}")
print(f"\nFirst 3 samples:\n{output[:, :3]}")
print("✓ Linear layer complete!")
```
