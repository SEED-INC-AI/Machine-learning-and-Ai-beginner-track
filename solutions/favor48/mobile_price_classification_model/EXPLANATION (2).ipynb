<h1> Explanation of the steps

<h2>1. Import libraries → Tools for data handling, plotting, and machine learning.

2. Load datasets → train.csv has features + labels (price_range), test.csv has only features.

3. Filter classes → Keep only rows where price_range is 0 (basic) or 1 (premium).

4. Split data → Features (X) are all columns except price_range; labels (y) are the price_range values.

5. Train model → Logistic Regression learns the relationship between features and whether a phone is basic or premium.

6. Predict → Apply the model to the test set to classify phones as 0 or 1.

7. Evaluate → Use training data to check accuracy, precision, recall, and confusion matrix.

8. Visualize → Bar chart shows how many phones are basic vs premium in the training set.


<h1>What the Model does

<h2>1. Logistic Regression calculates the probability that a phone belongs to class 0 (basic) or 1 (premium) based on its specs.

2. It uses a sigmoid function to map values between 0 and 1.

3. Phones with probability < 0.5 are classified as basic, ≥ 0.5 as premium.
