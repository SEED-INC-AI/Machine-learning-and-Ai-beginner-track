<h1>Exercise 08:Handling Missing Values

```python
df = pd.read_excel("network_usage.xlsx", engine="openpyxl")
# Reads the Excel file into a DataFrame.

print(df.isnull().sum())
# Shows how many missing values each column has.

df_dropped = df.copy()               # Make a safe copy
df_dropped.dropna(inplace=True)      # Remove any row with at least one NaN

df_filled = df.copy()                # Another safe copy
df_filled.fillna(df.select_dtypes(include="number").mean(), inplace=True)
# Fills missing values only in numeric columns with their mean

print("Original shape:", df.shape)       # Rows & columns in original dataset
print("After dropping:", df_dropped.shape)  # After removing rows
print("After filling:", df_filled.shape)    # After filling with means
```

<h1>Exercise 12:Grouping & Aggregation 

```python
df = pd.read_csv("transactions.txt", sep=',') 
df.head()
# Convert to numeric, forcing errors to NaN if non-numeric values exist
df['amount'] = pd.to_numeric(df['amount'], errors='coerce')

# Drop rows where 'amount' could not be converted
df = df.dropna(subset=['amount'])

print("Data preview:\n", df.head())
print("\nData info:\n")
print(df.info())

# Grouping & Aggregation tasks ---
# 1. Count the number of transactions per customer
transactions_per_customer = df.groupby("customer_id").size()

# 2. Calculate the average and total transaction amount per region
region_stats = df.groupby("region")["amount"].agg(["mean", "sum"])

# 3. Identify the region with the highest total revenue
top_region = df.groupby("region")["amount"].sum().idxmax()

# Display results ---
print("\nTransactions per customer:\n", transactions_per_customer)
print("\nAverage and total transaction amount per region:\n", region_stats)
print("\nRegion with highest total revenue:", top_region)
```

<h1>Execrcise 14:Data Merging

```python
#Load the datasets ---
customers = df = pd.read_csv("customers.csv")
transactions = df = pd.read_csv('transactions.txt', sep=',')

# Ensure customer_id is consistent in both datasets
customers['customer_id'] = pd.to_numeric(customers['customer_id'], errors='coerce')
transactions['customer_id'] = pd.to_numeric(transactions['customer_id'], errors='coerce')

# Drop rows with invalid customer_id
customers = customers.dropna(subset=['customer_id'])
transactions = transactions.dropna(subset=['customer_id'])

#Perform merges ---
# 1. Inner join: only customers with matching transactions
inner_merge = pd.merge(customers, transactions, on='customer_id', how='inner')

# 2. Left join: all customers, even those without transactions
left_merge = pd.merge(customers, transactions, on='customer_id', how='left')

# 3. Right join: all transactions, even if customer not in customers table
right_merge = pd.merge(customers, transactions, on='customer_id', how='right')

# Compare row counts ---
print("=== EXERCISE 14 RESULTS ===")
print("\nRow counts:")
print("Customers:", len(customers))
print("Transactions:", len(transactions))
print("Inner join:", len(inner_merge))
print("Left join:", len(left_merge))
print("Right join:", len(right_merge))

# --- Step 4: Explain differences ---
print("\nExplanation:")
print("- Inner join includes only customers who have transactions.")
print("- Left join includes all customers, with NaN for those without transactions.")
print("- Right join includes all transactions, with NaN for those without matching customers.")
print("- The difference between left and inner join row counts = number of customers with no transactions.")
```

<h1>Exercise 15:Left Anti Join(Advanced)

```python
 #Load the datasets ---
customers = df = pd.read_csv("customers.csv")
transactions = df = pd.read_csv('transactions.txt', sep=',')


# Ensure customer_id is numeric in both datasets
customers['customer_id'] = pd.to_numeric(customers['customer_id'], errors='coerce')
transactions['customer_id'] = pd.to_numeric(transactions['customer_id'], errors='coerce')

customers = customers.dropna(subset=['customer_id'])
transactions = transactions.dropna(subset=['customer_id'])

# Perform left join with indicator ---
merged = pd.merge(customers, transactions, on='customer_id', how='left', indicator=True)

#  Filter to keep only left_only records (customers with no transactions) ---
inactive_customers = merged[merged['_merge'] == 'left_only'].copy()

# Drop the helper column ---
inactive_customers.drop(columns=['_merge'], inplace=True)

# Count inactive customers ---
inactive_count = len(inactive_customers)

#  Display results ---
print("=== EXERCISE 15 RESULTS ===")
print("\nInactive customers (no transactions):\n", inactive_customers)
print("\nNumber of inactive customers:", inactive_count)
```
